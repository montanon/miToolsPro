# SankeyPlotter Demo

This notebook demonstrates key functionalities of the `SankeyPlotter` object.

```python
from mitoolspro.plotting.plots.sankey_plotter import SankeyNode, SankeyColumn, SankeyLink, SankeyPlotter
```

## 1. Create a Sankey Diagram by Passing Only Nodes

Provide nodes and auto connect all columns. The `connect_columns`handles the automatic connnection of nodes with corresponding names between consecutive columns and add a `SankeySinkNode` wherever necessary.

```python
columns = [
    SankeyColumn(name="First", period=1, nodes=[SankeyNode(name="A", count=10, period=1, rank=1),
    SankeyNode(name="B", count=20, period=1, rank=2),
    SankeyNode(name="C", count=30, period=1, rank=3),
    ]),
    SankeyColumn(name="Second", period=2, nodes=[SankeyNode(name="A", count=40, period=2, rank=1),
    SankeyNode(name="F", count=50, period=2, rank=2),
    SankeyNode(name="C", count=60, period=2, rank=3),
    ]),
    SankeyColumn(name="Third", period=3, nodes=[SankeyNode(name="C", count=70, period=3, rank=1),
    SankeyNode(name="D", count=80, period=3, rank=2),
    SankeyNode(name="A", count=90, period=3, rank=3),
    ]),
    SankeyColumn(name="Fourth", period=4, nodes=[SankeyNode(name="M", count=70, period=4, rank=1),
    SankeyNode(name="C", count=80, period=4, rank=2),
    SankeyNode(name="D", count=90, period=4, rank=3),
    ]),
]

diagram1 = SankeyPlotter(columns=columns)
diagram1.connect_columns()
diagram1.render(width=800, height=400)
```

## 2. Create a Sankey Diagram by Passing Nodes and Links

Provide nodes and link to specify the Sankey structure. The `create_sink_links` handles the automatic inclusion of `SankeySinkNode` wherever necessary.

```python
columns = [
    SankeyColumn(name="First", period=1, nodes=[SankeyNode(name="A", count=10, period=1, rank=1),
    SankeyNode(name="B", count=20, period=1, rank=2),
    SankeyNode(name="C", count=30, period=1, rank=3),
    ]),
    SankeyColumn(name="Second", period=2, nodes=[SankeyNode(name="A", count=40, period=2, rank=1),
    SankeyNode(name="F", count=50, period=2, rank=2),
    SankeyNode(name="C", count=60, period=2, rank=3),
    ]),
    SankeyColumn(name="Third", period=3, nodes=[SankeyNode(name="C", count=70, period=3, rank=1),
    SankeyNode(name="D", count=80, period=3, rank=2),
    SankeyNode(name="A", count=90, period=3, rank=3),
    ]),
    SankeyColumn(name="Fourth", period=4, nodes=[SankeyNode(name="M", count=70, period=4, rank=1),
    SankeyNode(name="C", count=80, period=4, rank=2),
    SankeyNode(name="D", count=90, period=4, rank=3),
    ]),
]
links = [
    SankeyLink(columns[0].nodes[0], columns[1].nodes[1], value=columns[0].nodes[0].count),
    SankeyLink(columns[0].nodes[0], columns[2].nodes[1], value=columns[0].nodes[0].count),
    SankeyLink(columns[0].nodes[0], columns[3].nodes[1], value=columns[0].nodes[0].count),
    SankeyLink(columns[0].nodes[1], columns[1].nodes[0], value=columns[0].nodes[1].count),
    SankeyLink(columns[0].nodes[2], columns[1].nodes[2], value=columns[0].nodes[2].count),
    SankeyLink(columns[1].nodes[0], columns[2].nodes[2], value=columns[1].nodes[0].count),
    SankeyLink(columns[1].nodes[2], columns[2].nodes[0], value=columns[1].nodes[2].count),
    SankeyLink(columns[2].nodes[0], columns[3].nodes[2], value=columns[2].nodes[0].count),
    SankeyLink(columns[2].nodes[1], columns[3].nodes[1], value=columns[2].nodes[1].count),
    SankeyLink(columns[2].nodes[2], columns[3].nodes[0], value=columns[2].nodes[2].count),
]
diagram2 = SankeyPlotter(columns=columns, links=links)
diagram2.render(width=800, height=400)
```

Column 2 `"F"`nodes is not connected to anything and a novel `"D"`node appears in Column 3. We can add a `SankeySinkNode` to connect the `"F"`node to the `"D"`node to show that something comes from "outside" the Sankey Diagram.

```python
diagram2 = SankeyPlotter(columns=columns, links=links)
diagram2.create_sink_links()
diagram2.render(width=800, height=400)
```

## 3. Store and Load Sankey Diagram

The `SankeyPlotter`object has functionality to store and load the diagrams in human-readable structured `JSON` format or `DataFrames`.

#### To/From JSON

```python
diagram_json = diagram2.to_json()
print(diagram_json)
```

```python
diagram3 = SankeyPlotter.from_json(diagram_json)
diagram3.render(width=800, height=400)
```

#### To/From DataFrames

This functionality returns a `DataFrame`with the nodes data as default. A flag can be set to return the links as well. The `SankeyPlotter`object can be initialized from nodes or a pair of nodes and links.

```python
nodes,links = diagram2.to_dataframe()
```

```python
nodes.head()
```

```python
links.head()
```

If links are not provided, the `SankeyPlotter`object will be initialized with `connect_columns` if the `auto_link`flag is kept as `True`. Otherwise, the `SankeyPlotter`object will be initialized with the provided nodes only and links must be managed manually.

```python
diagram4 = SankeyPlotter.from_dataframe(nodes, None, auto_link=True)
diagram4.render(width=800, height=400)
```

```python
diagram4 = SankeyPlotter.from_dataframe(nodes, None, auto_link=False)
diagram4.connect_columns()
diagram4.render(width=800, height=400)
```

If links are provided with the corresponding nodes, the `SankeyPlotter`object will be initialized with the provided structure.

```python
diagram4 = SankeyPlotter.from_dataframe(nodes, links)
diagram4.render(width=800, height=400)
```
