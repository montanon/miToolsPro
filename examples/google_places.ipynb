```python
from mitoolspro.google_utils.places.workflow import PlacesSamplingWorkflow
from mitoolspro.google_utils.places.client import GooglePlacesClient
from pathlib import Path
import os
```

```python
geojson_path = "./Santiago_Comunas.geojson"
```

```python
workflow = PlacesSamplingWorkflow(
        city_name="Santiago",
        geojson_path=geojson_path,
        project_folder=Path("Here"),
        client=GooglePlacesClient(api_key=os.getenv("GOOGLE_PLACES_API_KEY")),
        show=True,
        recalculate=True,
    )
```

```python
workflow.run()
```
